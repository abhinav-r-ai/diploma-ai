```python
q1=data['Applicant_Income'].quantile(0.25)
q3=data['Applicant_Income'].quantile(.75)

print(f"q1={q1}")
print(f"q3={q3}")
```

```python
IQR=q3-q1
print(f"IQR={IQR}")
```

```python
ub=q3+(1.5*IQR)
lb=q3-(1.5*IQR)


print(f"Upper Bound = {ub}")
print(f"Lower Bound = {lb}")
```

```python

ol=data[(data['Applicant_Income']>ub)|(data['Applicant_Income']<lb)]
display(ol)
```
