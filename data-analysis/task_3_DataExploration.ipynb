```python
a=['Hours Studied', 'Previous Scores','Sleep Hours', 'Extracurricular Activities','Sample Question Papers Practiced', 'Performance Index']
```

```python

for i in a:
  if i!='Extracurricular Activities':
    print(f"mean of {i} = {data[i].mean()}")

    print(f"median of {i}={data[i].median()}")

    print(f"mode of {i}={data [i].mode()[0]}")
    print("="*30)
  else:
    print(f"mode of {i}={data [i].mode()[0]}")
```

```python
b=['Hours Studied', 'Previous Scores','Sleep Hours','Sample Question Papers Practiced', 'Performance Index']
```

```python
for i in b:
  print(f'range of {i}={data[i].max()-data[i].min()}')
  print(f'std of {i}={data[i].std()}')
  print(f'variance of {i}={data[i].var()}')

  q1 = data[i].quantile(0.25)
  q3 = data[i].quantile(0.75)
  iqr = q3 - q1

  print(f'iqr of {i}={iqr}')
  print(f'skewness of {i}={data[i].skew()}')
  print(f' kurtosis of {i}={data[i].kurtosis()}')
  print("="*65)


```
